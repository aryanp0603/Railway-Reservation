# Stone Paper Scissor Game

```python
player1= input("Enter Stone, Paper or Scissor".lower())
player2 = input("Enter Stone, Paper or Scissor".lower())

if player1 == player2:
    print("It's a Draw.....!")
elif player1 == "stone":
    if player2 == "scissor":
        print("Player 1 won....!")
    else:
        print("Player 2 won....!")
elif player1 == "paper":
    if player2 == "stone":
        print("Player 1 won....!")
    else:
        print("Player 2 won....!")
elif player1 == "scissor":
    if player2 == "paper":
        print("Player 1 won....!")
    else:
        print("Player 2 won....!")
else:
    print("Invalid Input!!!")
```
