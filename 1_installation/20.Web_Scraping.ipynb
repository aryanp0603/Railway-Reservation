```python
import requests
from bs4 import BeautifulSoup

url = "https://contentitvedant.github.io/Railway_Booking/trains.html"

response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

table = soup.find(id="train-schedule-table")
rows = table.find("tbody").find_all("tr")

print("Train Schedule:")
print("=" * 60)

for row in rows:
    cells = row.find_all("td")
    train_no = cells[0].get_text()
    train_name = cells[1].get_text()
    from_stn = cells[2].get_text()
    to_stn     = cells[3].get_text()
    departure  = cells[4].get_text()
    arrival    = cells[5].get_text()
    duration   = cells[6].get_text()

    print(train_no, "|", train_name)
    print("   From:", from_stn, "-> To:", to_stn)
    print("   Departure:", departure, "| Arrival:", arrival, "| Duration:", duration)
    print("=" * 60)
```

Task 6

```python

import requests
from bs4 import BeautifulSoup
import csv


class RailwayScraper:

    def __init__(self):
        self.base_url = "https://contentitvedant.github.io/Railway_Booking/"

    def get_soup(self, page):
        url = self.base_url + page
        response = requests.get(url)

        soup = BeautifulSoup(response.text, "html.parser")

        print("Fetched:", url, "| Status:", response.status_code)

        return soup

    def get_train_schedule(self):

        soup = self.get_soup("trains.html")

        table = soup.find(id="train-schedule-table")

        rows = table.find("tbody").find_all("tr")

        trains = []

        for row in rows:

            cells = row.find_all("td")

            trains.append({

                "train_no": cells[0].get_text(strip=True),

                "train_name": cells[1].get_text(strip=True),

                "from": cells[2].get_text(strip=True),

                "to": cells[3].get_text(strip=True),

                "departure": cells[4].get_text(strip=True),

                "arrival": cells[5].get_text(strip=True),

                "duration": cells[6].get_text(strip=True)

            })

        return trains

    def get_passengers(self):

        soup = self.get_soup("passengers.html")

        table = soup.find(id="passenger-records-table")

        rows = table.find("tbody").find_all("tr")

        passengers = []

        for row in rows:

            passengers.append({

                "pnr": row["data-pnr"],

                "name": row.find(class_="passenger-name").get_text(strip=True),

                "date": row.find(class_="journey-date").get_text(strip=True),

                "class": row.find(class_="travel-class").get_text(strip=True),

                "seat": row.find(class_="seat-number").get_text(strip=True),

                "fare": row.find(class_="fare").get_text(strip=True)

            })

        return passengers

    def get_seat_availability(self):

        soup = self.get_soup("seat-availability.html")

        table = soup.find(id="seat-availability-table")

        rows = table.find("tbody").find_all("tr")

        available = []

        for row in rows:

            available.append({

                "train_no": row["data-train"],

                "train_name": row.find(class_="train-name").get_text(strip=True),

                "seats_sl": row.find(class_="seats-sl").get_text(strip=True),

                "seats_3a": row.find(class_="seats-3a").get_text(strip=True),

                "seats_2a": row.find(class_="seats-2a").get_text(strip=True),

                "seats_1a": row.find(class_="seats-1a").get_text(strip=True)

            })

        return available
```

```python

# Create scraper object

scraper = RailwayScraper()

# Get train schedule

train = scraper.get_train_schedule()

# Display train schedule

print("\n--- TRAIN SCHEDULE ---")

for t in train:

    print(

        t["train_no"], "|",

        t["train_name"], "|",

        t["from"], "->", t["to"], "|",

        t["duration"]

    )
```

```python

# Save train schedule in TXT file

with open("train_schedule.txt", "w", encoding="utf-8") as file:

    for t in train:

        file.write(

            t["train_no"] + " - " +

            t["train_name"] + "\n"

        )

print("Train schedule saved successfully.")
```

```python

# Read TXT file

with open("train_schedule.txt", "r", encoding="utf-8") as file:

    print("Stored Train Schedule")

    print("=" * 40)

    print(file.read())
```

```python

# Create CSV file

with open("train_schedule.csv", "w", newline="", encoding="utf-8") as file:

    writer = csv.writer(file)

    writer.writerow([

        "Train Number",

        "Train Name",

        "From",

        "To",

        "Departure",

        "Arrival",

        "Duration"

    ])

    for t in train:

        writer.writerow([

            t["train_no"],

            t["train_name"],

            t["from"],

            t["to"],

            t["departure"],

            t["arrival"],

            t["duration"]

        ])

print("CSV file created successfully.")
```
