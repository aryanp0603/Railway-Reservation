using for loop count of vowels

```python
a = "ITvedant"
vowel = "AEIOUaeiou"
count = 0
for i in a:
    if i in vowel:
        count = count + 1
print(count)
```

```python
a = input("Enter your word:")
b = "AEIOUaeiou"
if a[0] in b:
    print("An", a)
else:
    print("A", a)
```

Create Passenger Booking Function

```python
def book_ticket(passenger_name, train_name):
    print("Passenger Name:", passenger_name)
    print("Train Name:", train_name)
    print("Reservation Status: Confirmed!!!")
book_ticket("Rahul Sharma", "Rajdhani Express")
```

Create Fare Caluclation Function

```python
def Total_ticket_fare(ticket_fare, passengers):
    print("Total ticket fare:", ticket_fare * passengers)
Total_ticket_fare(3600, 4)
```

Create passenger verification function

```python
def verify_passenger(age):
    if age >= 18:
        print("Eligible for Reservation")
    else:
        print("Not Eligible")
verify_passenger(29)
```

Build Ticket generation function

```python
def generated_ticket(passenger_name, train_name, seat_number):
    print("===================================================")
    print("Railway Reservation Ticket")
    print("===================================================")
    print("Passenger Name:", passenger_name)
    print("Train Name:", train_name)
    print("Seat number", seat_number)
generated_ticket("Rahul Sharma", "Rajdhani Express", "B2-51")
```
