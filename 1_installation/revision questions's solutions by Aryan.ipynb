```python
print("Creating Variable")
name = "Aryan Patel"
print("Name:", name)
print("\n Name in UPPER CASE")
name1 = name.upper()
print(name1)
print("\nNAme in LOWER CASE")
name2 = name.lower()
print(name2)
print("\nName with only the first letter of each word capitalised")
name3= name.format()
print(name3)
```

Task 6

```python
print("Creating One String")
string = "   I love Python   "
print("String:", string)
print("\nRemoving the extra spaces from both side(beginning and end)")
string1 = string.strip()
print(string1)
print("\nReplace the word 'love' with 'enjoy' ")
string2 = string1.replace("love", "enjoy")
print(string2)
print("\nSplit the sentence into list")
string3 = string2.split()
print(string3)
```
