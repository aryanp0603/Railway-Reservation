```python
class car:
    def __init__(self):
        print("Ye car ka constructor hai")
    def display(self):
        print("Your car is awesome")
class AutomaticCar(car):
    def __init__(self):
        print("Ye automatic car ka constructor hai")
        super().__init__()
    def display(self):
        print("Automatic Car is awesome too.")
a = AutomaticCar()
a.display()
```

```python
def decorate(func):
    def wrap():
        print("Happy Birthday!!!")
        func()
    return wrap

@decorate
def bday():
    a = input("Enter a name:")
    # print(f"Happy Birthday {a}.....")
    print(a)
bday()
```
