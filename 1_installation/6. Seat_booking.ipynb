While Loop

```python
available_seats = int(input("Enter Number of seats:"))

while available_seats >= 0 :
    print("Seat Booked Successfully....")
    print("Remaining seats :", available_seats)
    available_seats = available_seats - 1
print("All seats are filled......")
```

For Loop

```python
print("Available Seat Numbers")
for i in range(1, 11):
    print("Seat Number:",i)
```

Stop Booking When Seats Become Full

```python
total_seats= 5

for booking in range(1, 11):
    if booking > total_seats:
        print("No Seats Available")
        break
    if booking == 3:
        print("Checking Special Reservation")
        continue
    pass
    print("Seats", booking, "Booked Successfully")
```
