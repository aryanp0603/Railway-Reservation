```python
a = (1, 2, 3, 1, 3, 1, 1, 4)
Unique = []
for i in a:
    if a.count(i) > 1:
        if i in Unique:
            continue
        else:
            Unique.append(i)
    else:
        Unique.append(i)
print(tuple(Unique))
```
