Task 2: Perform operations on List

Create List

```python
passenger = ["Rahul Sharma", "Priya Patel", "Amit Verma"]
```

```python
print(passenger)
```

Add New Passenger Using append()

```python
passenger.append("Sneha Joshi")
print(passenger)
```

Remove Passenger Using remove()

```python
passenger.remove("Amit Verma")
print(passenger)
```

Insert Passenger using insert()

```python
passenger.insert(1, "Rohan Mehta")
print(passenger)
```

Remove item using pop()

```python
passenger.pop(2)
print(passenger)
```

Display Updated List

```python
print("\nUpdated Reservation List:")
print(passenger)
```
