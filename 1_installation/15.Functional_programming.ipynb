Create Lambda Function for Fare Discount

```python
discount_fare = lambda fare : fare-200
print("Discounted Fare:", discount_fare(2500))
```

Apply map() Function for Fare Calculation

```python
ticket_fare = [2500, 650, 3060, 4274]
updated_fare = list(
    map(
        lambda fare : fare + 100, 
        ticket_fare
    )
)

print(updated_fare)
```

Apply filter() Function for confirmed tickets

```python
reservation_status = [
    "Confirmed",
    "Waiting",
    "Confirmed",
    "RAC"
]

confirmed_ticket = list(
    filter(
        lambda status : status == "Confirmed",
        reservation_status
    )
)

print(confirmed_ticket)
```

Combine Passenger and Seat Data Using zip()

```python
passenger = [
    "Rahul Sharma",
    "Priya Singh",
    "Amit Verma"
]

Seats = [
    "B1-21",
    "B1-22",
    "B1-23"
]

reservation_data = list(
    zip(passenger, Seats)
)

print(reservation_data)
```

Apply enumerate() Function

```python
passneger_list = [
    "Rahul Sharma",
    "Priya Singh",
    "Amit Verma"
]

for index, passenger in enumerate(passneger_list):
    print(index, passenger)
```

```python
ticket_fare = [2500, 800, 3200, 600]

valid_fares = list(
    filter(
        lambda fare : fare > 1000, ticket_fare
    )
)

final_fares = list(
    map(
        lambda fare : fare + 100, valid_fares
    )
)

print(final_fares)
```

```python
passenger = ["Rahul", "Priya", "Amit"]
seats = ["S1", "S2", "S3"]
fares = [2500, 1800, 3200]

reservation_data = list(
    zip(passenger, seats, fares)
)

print("===================================")
print(" Railway Reservation Report ")
print("===================================")

for index, data in enumerate(reservation_data):

   passenger, seat, fare = data

   final_fare = lambda x: x - 100

   print("Reservation ID :", index + 1)
   print("Passenger :", passenger)
   print("Seat :", seat)
   print("Final Fare :", final_fare(fare))
   print("-----------------------------------")
```
