Take input from user and check whether the number is prime or not

```python
def prime_number(n):
    n =  int(input("Enter number:"))
    i = 2
    while i < n:
        if n % i == 0:
            print("Not Prime")
            break
        i = i + 1

    else:
        if n > 1:
            print("Prime")
        else:
            print("Not prime")

prime_number(n)
```
